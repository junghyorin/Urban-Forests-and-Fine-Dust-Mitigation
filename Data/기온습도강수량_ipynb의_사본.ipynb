```python
import pandas as pd

# CSV 파일 경로
file_path = '/content/기상청_온도습도기온.csv'

# CSV 파일 불러오기
df = pd.read_csv(file_path, encoding='euc-kr')

print(df.tail(100))

```

```python
# '일시' 컬럼을 datetime 형식으로 변환
df['일시'] = pd.to_datetime(df['일시'])

# 필터 조건 설정
valid_weekdays = [0, 2, 4, 5]  # 월(0), 수(2), 금(4), 토(5)
valid_hours = [3, 8, 14, 18]   # 03시, 08시, 14시, 18시

# 조건에 맞는 데이터만 필터링
filtered_df = df[
    (df['일시'].dt.weekday.isin(valid_weekdays)) &
    (df['일시'].dt.hour.isin(valid_hours)) &
    (df['일시'] >= pd.to_datetime('2024-01-01'))
]

# 결과 확인
print(filtered_df.head())
```

```python
filtered_df.to_csv("filtered_all.csv", index=False)
```

```python
import pandas as pd

# 파일 경로
file_path = '/content/final_pm10_with_urbanforest-2.csv'

# CSV 파일 불러오기
df2 = pd.read_csv(file_path)

```

```python
# '일시' 컬럼을 datetime 형식으로 변환
df['ymdt'] = pd.to_datetime(df['일시'])

# 'ymdt'를 원하는 형식(YYYYMMDDHHMM)으로 변환
df['ymdt'] = df['ymdt'].dt.strftime('%Y%m%d%H%M')

# 변환된 'ymdt' 확인
print(df['ymdt'].head())
```

```python
print("pm_df 시간대 예시:")
print(df2['ymdt'].drop_duplicates().sort_values().head(5))

print("\nweather_df 시간대 예시:")
print(df['ymdt'].drop_duplicates().sort_values().head(5))
```

```python
print(df2['station_name'].unique())
print(df['station_name'].unique())

print(df2['ymdt'].sort_values().unique()[:5])
print(df['ymdt'].sort_values().unique()[:5])
```

```python
# 1. datetime으로 정확히 변환 (int → str → datetime)
df2['ymdt'] = pd.to_datetime(df2['ymdt'].astype(str), format='%Y%m%d%H%M')
df['ymdt'] = pd.to_datetime(df['ymdt'].astype(str), format='%Y%m%d%H%M')

# 2. 지점명 컬럼 통일 ('지점명' → 'station_name') 및 '구' 추가
df = df.rename(columns={'지점명': 'station_name'})
df['station_name'] = df['station_name'].str.strip().apply(lambda x: x + '구' if '구' not in x else x)

# 3. 병합: df2 기준으로 df를 붙이기
merged_df = pd.merge(df2, df, on=['ymdt', 'station_name'], how='left')

# 4. 결과 확인
print(merged_df.head())

```

```python
# 병합 이후 df 기준 컬럼이 중복되었을 가능성 있는 것들 제거
merged_df = merged_df.drop(columns=['일시', '지점'], errors='ignore')
```

```python
print(merged_df.head(100))
```

```python
# 병합된 데이터를 CSV로 저장
merged_df.to_csv('merged_weather_pm_data_final_null_o.csv', index=False)
```

```python
print(merged_df.isnull().sum())
```

```python

# 연속형: 평균으로 대체 (기온, 습도, 기압, 풍속)
for col in ['기온(°C)', '습도(%)', '현지기압(hPa)', '풍속(m/s)']:
    merged_df[col].fillna(merged_df[col].mean(), inplace=True)

# 강수량: 비가 안 왔다고 가정하여 0으로 대체
merged_df['강수량(mm)'].fillna(0, inplace=True)

# 미세먼지 처리 - 같은 시간대 평균으로 채우기
merged_df['hour'] = merged_df['ymdt'].dt.hour
merged_df['pm10'] = merged_df.groupby('hour')['pm10'].transform(lambda x: x.fillna(x.mean()))


# 풍향: 각도 데이터 → 평균보다 중앙값으로 대체 (또는 최빈값도 가능)
merged_df['풍향(deg)'].fillna(merged_df['풍향(deg)'].median(), inplace=True)
print(merged_df.isnull().sum())

```

```python
merged_df.to_csv('merged_weather_pm_data_null_x.csv', index=False)
```
